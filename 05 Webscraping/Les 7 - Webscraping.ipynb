```python
import requests
from bs4 import BeautifulSoup

headers={'User-Agent': 'Mozilla/5.0'} #doen alsof je een browser bent
response = requests.get("https://nl.wikipedia.org/wiki/Programmeertaal", headers=headers)

soup = BeautifulSoup(response.text)
a_tags = soup.find_all('a', attrs={"rel": "mw:WikiLink"})
print(a_tags)

# <a rel="mw:WikiLink" href="//nl.wikipedia.org/wiki/Taal" title="Taal" id="mwDg">natuurlijke talen</a>

for tag in a_tags:
    print(tag.get_text())
```

In plaats van find_all() kan je ook .select() en .select_one() gebruiken. Deze methodes werken met CSS-selectors — dezelfde syntax die je ziet in F12!

```python
import requests
from bs4 import BeautifulSoup

headers={'User-Agent': 'Mozilla/5.0'} #doen alsof je een browser bent
response = requests.get("https://nl.wikipedia.org/wiki/Programmeertaal", headers=headers)

soup = BeautifulSoup(response.text)
tag = soup.select("#mwMQ")
print(tag)
```

## Beelden scrapen



```python
import requests
from bs4 import BeautifulSoup

headers={'User-Agent': 'Mozilla/5.0'} #doen alsof je een browser bent
response = requests.get("https://www.meteoblue.com/nl/weer/webcams/parijs_frankrijk_2988507", headers=headers)

soup = BeautifulSoup(response.text)
img_tags = soup.find_all("img")

count = 0

for img in img_tags:
    img_url = img["src"]
    print(img_url)

    if img_url.endswith(".jpg"):
        response_img = requests.get(img_url)
        # Create image-file --> 'wb' = write binary mode (essential for non-text files like JPG/PNG)
        with open(f"img/meteoblue_{count}.jpg", 'wb') as file:
            file.write(response_img.content) # .content is used for raw bytes
        count += 1
```

## Dynamische pagina's
https://pypi.org/project/playwright/

requests downloadt enkel de HTML — JavaScript wordt nooit uitgevoerd. Playwright start een échte browser op en stuurt die aan via Python.

Installatie: pip install playwright en python -m playwright install chromium

page.goto("url") → navigeer naar een pagina

page.query_selector_all("span.text") → selecteer elementen (zelfde CSS-selectors als BeautifulSoup!)

element.inner_text() → lees de tekst uit

page.click("li.next a") → klik op een element

page.fill("input#username", "user") → typ tekst in een inputveld

page.wait_for_load_state("networkidle") → wacht tot de pagina volledig geladen is
