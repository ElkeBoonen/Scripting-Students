## Scrapen data in database

https://yougov.com/en-us/ratings/all-time-actors-actresses 


scrapen top 100 actors naar actors-tabel

```python
# stap 1 : scrapen --> zorg dat je uberhaupt de data te pakken krijgt!!!!

import requests
from bs4 import BeautifulSoup

url = "https://yougov.com/en-us/ratings/all-time-actors-actresses"
headers = {'User-agent': 'Mozilla/5.0'} 

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text)

# <span _ngcontent-yg-app-c318061963="" class="entity-name multi-word">Betty White</span>

data = []
actors = soup.find_all("span",attrs={"class":"entity-name multi-word"})
for actor in actors:
    print(actor.getText()) # of contents
    data.append(actor.contents) # we gebruiken hier contents omdat de data bij execute query een tuple of list verwacht!!!

print(data)
```

```python
import mysql.connector

db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="movies_db"
)

cursor = db.cursor()

query = "INSERT INTO actors (name) VALUES (%s)"
for actor in data:
    cursor.execute(query, actor)

db.commit() # SUPER BELANGRIJK!!!!

cursor.close()
db.close()
```
