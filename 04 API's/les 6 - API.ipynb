## Herhaling
Je wilt een overzicht maken van alle familienamen in de Provincie Antwerpen, maar je wilt enkel de namen die meer dan 500 keer voorkomen. Omdat de data "vervuild" is met aanhalingstekens en komma's in de getallen (bijv. "1,234"), moet je die eerst opschonen.

- Maak een map Provincie_Rapporten aan.
- Lees het bestand in. Sla de eerste twee regels over (de titel en de header).
- Filter op de vierde kolom (Provincie/Gewest) voor "Provincie Antwerpen".
- Zet het aantal (laatste kolom) om naar een echt getal (verwijder " en ,).
- Schrijf de namen die aan de voorwaarde voldoen naar Antwerpen_500plus.csv.

```python
import re,os, pandas as pd

if not os.path.exists("Provincie_Rapporten"): # parameters bij mkdir kan ook!!
    os.mkdir("Provincie_Rapporten")

file = pd.read_csv("Fam_names_STATBEL_prov_nl(2025).csv", encoding="ISO 8859-1", skiprows=1)
antwerpen = file[(file["Provincie/Gewest"]=="Provincie Antwerpen")]
antwerpen["Aantal"] = antwerpen["Aantal"].str.replace(",","").astype(int)
antwerpen = antwerpen[(antwerpen["Aantal"] >= 500)]

print(antwerpen)

antwerpen.to_excel("Provincie_Rapporten/antwerpen.xlsx")
```

```python
import requests

url = "https://the-love-calculator.p.rapidapi.com/love-calculator"

fname = input("Geef de eerste naam")
sname = input("Geef de tweede naam")

querystring = {"fname":fname,"sname":sname}

headers = {
	"x-rapidapi-key": "",
	"x-rapidapi-host": "the-love-calculator.p.rapidapi.com",
	"Content-Type": "application/json"
}

response = requests.get(url, headers=headers, params=querystring)

print(response.json())
```

```python
import requests

url = "https://netflix54.p.rapidapi.com/search/"

querystring = {"query":"stranger things","offset":"0","limit_titles":"50","limit_suggestions":"20","lang":"en"}

headers = {
	"x-rapidapi-key": "",
	"x-rapidapi-host": "netflix54.p.rapidapi.com",
	"Content-Type": "application/json"
}

response = requests.get(url, headers=headers, params=querystring)

print(response.json())

for title in response.json()["titles"]:
    print(title)
```
